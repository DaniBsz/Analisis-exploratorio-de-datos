```python
#Calculo de la densidad poblacional
df['density'] = df['population'] / df['area']
```

```python
#Calculo de la densidad de usuarios de internet
df['IntuDensity']= df['internet_users'] / df['population']
```

```python
df[df['IntuDensity'] == df['IntuDensity'].max()]['country']
```

```python
df['internet_users_per_sqkm'] = df['internet_users'] / df['area']
print(f"País con la mayor cantidad de usuarios de internet por kilómetro cuadrado: {df[df['internet_users_per_sqkm'] == df['internet_users_per_sqkm'].max()]['country'].iloc[0]}")
```

```python
df['percentage_internet_users'] = (df['internet_users'] / df['population']) * 100
print(f"País con el mayor porcentaje de usuarios de internet: {df[df['percentage_internet_users'] == df['percentage_internet_users'].max()]['country'].iloc[0]}")
```
